```python
import numpy as np
import pandas as pd
```

```python
df_locations = pd.read_csv("Sources\ech_annonces_locations_68.csv", on_bad_lines="skip", sep=";")
df_locations.head()
```

```python
df_ventes = pd.read_csv("Sources\ech_annonces_ventes_68.csv", sep=";")
df_ventes.head()
```

```python
df = pd.concat([df_locations, df_ventes], axis=0)
df.head()
```

```python
df['etage'] = df['etage'].replace([999, 99], df['etage'].median())
df['etage'].value_counts()
```

```python
df.loc[(df['parking'].isna()) & (df['places_parking'].notna()), 'parking'] = True
```
