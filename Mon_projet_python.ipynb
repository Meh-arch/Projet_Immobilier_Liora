```python


from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
my_df['ges_class'] = le.fit_transform(df['ges_class'])
my_df['dpeL'] = le.fit_transform(df['dpeL'])
my_df['chauffage_energie'] = le.fit_transform(df['chauffage_energie'])
my_df['chauffage_systeme'] = le.fit_transform(df['chauffage_systeme'])
my_df['chauffage_mode'] = le.fit_transform(df['chauffage_mode'])

df_Energies = my_df[['ges_class', 'dpeL', 'chauffage_energie', 'chauffage_systeme', 'chauffage_mode', 'dpeC']]

print(df_Energies)
```

```python
correlation_matrix = df_Energies.corr()
f = plt.figure(figsize=(10, 7))
plt.matshow(correlation_matrix, fignum=f.number)
cb = plt.colorbar()
cb.ax.tick_params(labelsize=14)
plt.show()
```
